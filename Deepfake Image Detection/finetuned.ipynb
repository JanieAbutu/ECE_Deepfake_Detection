```python
# Install torchinfo if not already installed
# pip install torchinfo

from torchvision import models
import torch
import torch.nn as nn
from torchinfo import summary

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

# ================================================================
# EfficientNet-B7 for binary classification
# ================================================================
class EfficientNetB7Binary(nn.Module):
    def __init__(self):
        super().__init__()
        # Load pretrained EfficientNet-B7
        self.backbone = models.efficientnet_b7(weights="IMAGENET1K_V1")

        # Replace classifier for binary output
        in_features = self.backbone.classifier[1].in_features
        self.backbone.classifier = nn.Sequential(
            nn.Dropout(0.5),
            nn.Linear(in_features, 1)  # binary output
        )

    def forward(self, x):
        return self.backbone(x)

# Instantiate model
model = EfficientNetB7Binary().to(DEVICE)

# Generate layer-wise summary
# input_size: (batch_size, channels, height, width)
summary(model, input_size=(1, 3, 224, 224))
```

```python
from torch import nn

def count_params(model):
    total_params = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"Total parameters: {total_params}")
    print(f"Trainable parameters: {trainable_params}\n")
    print(model)  # prints layer-by-layer structure

# Usage
model = EfficientNetB7Binary()
count_params(model)
```
